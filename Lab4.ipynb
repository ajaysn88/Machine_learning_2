```python
# ============================================================
# CANDIDATE-ELIMINATION ALGORITHM
# ROWS 2 TO 10 ONLY
# ============================================================

import pandas as pd
from google.colab import files

# ------------------------------------------------------------
# 1. Upload Dataset
# ------------------------------------------------------------

uploaded = files.upload()

data = pd.read_csv("play_tennis_dataset.csv")

# Take ONLY rows 2 to 10
data = data.iloc[1:20].copy()

print("Selected Rows (2 to 10):")
print(data)


# ------------------------------------------------------------
# 2. Handle Missing Values
# ------------------------------------------------------------

for column in data.columns:
    if data[column].isnull().sum() > 0:
        data[column] = data[column].fillna(
            data[column].mode()[0]
        )


# ------------------------------------------------------------
# 3. Select Attributes
# ------------------------------------------------------------

attributes = [
    'Outlook',
    'Temperature',
    'Humidity',
    'Wind'
]

target = 'Play'


# ------------------------------------------------------------
# 4. Candidate-Elimination Algorithm
# ------------------------------------------------------------

def covers(hypothesis, example):
    return all(
        hypothesis[i] == '?' or
        hypothesis[i] == example[i]
        for i in range(len(hypothesis))
    )


def candidate_elimination(data):

    # Most Specific Hypothesis
    S = ['0'] * len(attributes)

    # Most General Hypothesis
    G = [['?'] * len(attributes)]

    print("\n" + "=" * 60)
    print("CANDIDATE-ELIMINATION PROCESS")
    print("=" * 60)

    for count, (_, row) in enumerate(data.iterrows(), start=2):

        example = [row[attr] for attr in attributes]
        label = row[target]

        print(f"\nExample {count}")
        print("Attributes:", example)
        print("Class:", label)

        # ----------------------------------------------------
        # POSITIVE EXAMPLE
        # ----------------------------------------------------

        if label == 'Yes':

            # Remove G hypotheses that don't cover example
            G = [
                g for g in G
                if covers(g, example)
            ]

            # Generalize S
            for i in range(len(attributes)):

                if S[i] == '0':
                    S[i] = example[i]

                elif S[i] != example[i]:
                    S[i] = '?'

        # ----------------------------------------------------
        # NEGATIVE EXAMPLE
        # ----------------------------------------------------

        else:

            new_G = []

            for g in G:

                if covers(g, example):

                    # Specialize G
                    for i in range(len(attributes)):

                        if g[i] == '?':

                            if S[i] != '0' and S[i] != '?':
                                new_h = g.copy()
                                new_h[i] = S[i]

                                if not covers(new_h, example):
                                    new_G.append(new_h)

                else:
                    new_G.append(g)

            G = new_G

        print("S =", S)
        print("G =", G)

    return S, G


# ------------------------------------------------------------
# 5. Run Algorithm
# ------------------------------------------------------------

S, G = candidate_elimination(data)


# ------------------------------------------------------------
# 6. Final Version Space
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("FINAL VERSION SPACE")
print("=" * 60)

print("\nMost Specific Hypothesis (S):")
print("<", ", ".join(S), ">")

print("\nMost General Hypotheses (G):")

if len(G) == 0:
    print("No hypotheses")

else:
    for g in G:
        print("<", ", ".join(g), ">")
```
