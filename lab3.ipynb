```python
# Find-S Algorithm - Hardcoded Dataset

# Dataset
data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High',   'Weak', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High',   'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High',   'Strong', 'Cool', 'Change', 'Yes']
]

# Attribute names
attributes = [
    'Sky',
    'AirTemp',
    'Humidity',
    'Wind',
    'Water',
    'Forecast'
]

# Target column
target_index = 6

# Initialize hypothesis with the most specific hypothesis
# 'Ø' means no value has been assigned yet
hypothesis = ['Ø'] * len(attributes)

print("Initial Hypothesis:", hypothesis)

# Find-S algorithm
for row in data:

    # Process only positive examples
    if row[target_index] == 'Yes':

        print("\nProcessing Positive Example:", row[:-1])

        for i in range(len(attributes)):

            # If hypothesis is empty, copy the attribute value
            if hypothesis[i] == 'Ø':
                hypothesis[i] = row[i]

            # If values are different, generalize to '?'
            elif hypothesis[i] != row[i]:
                hypothesis[i] = '?'

        print("Current Hypothesis:", hypothesis)

# Final hypothesis
print("\n--------------------------------")
print("Final Hypothesis:")
print(hypothesis)
print("--------------------------------")
```

```python
import pandas as pd
from google.colab import files

# Upload CSV
uploaded = files.upload()

filename = list(uploaded.keys())[0]

# Read dataset
data = pd.read_csv(filename)

print("Original Dataset Shape:", data.shape)

# --------------------------------------------
# Use first 14 examples
# --------------------------------------------
data = data.head(14)

# Remove Day column
data = data.drop("Day", axis=1)

# Remove rows with missing values
data = data.dropna()

print("\nDataset used for Find-S:")
print(data)

# --------------------------------------------
# Find-S Algorithm
# --------------------------------------------

target = "Play"

attributes = ["Outlook", "Temperature", "Humidity", "Wind"]

# Most specific hypothesis
hypothesis = ["∅"] * len(attributes)

print("\nInitial Hypothesis:")
print(hypothesis)

# Process each example
for index, row in data.iterrows():

    # Find-S considers only positive examples
    if row[target] == "Yes":

        print("\nPositive Example:")
        print(list(row[attributes]))

        for i in range(len(attributes)):

            # First positive example
            if hypothesis[i] == "∅":
                hypothesis[i] = row[attributes[i]]

            # Different value -> generalize
            elif hypothesis[i] != row[attributes[i]]:
                hypothesis[i] = "?"

        print("Current Hypothesis:")
        print(hypothesis)

# --------------------------------------------
# Final Hypothesis
# --------------------------------------------

print("\n================================")
print("FINAL HYPOTHESIS")
print("================================")

for attribute, value in zip(attributes, hypothesis):
    print(attribute, "=", value)

print("\nFinal Hypothesis:")
print(hypothesis)
```
